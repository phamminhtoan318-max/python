```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv('bank.csv')

print("Kích thước bảng dữ liệu (shape):", df.shape)

print("\nDanh sách các cột:")
print(df.columns.tolist())
```

```python
# 1. Xem trước 5 dòng đầu tiên
print("--- 5 DÒNG ĐẦU TIÊN CỦA DỮ LIỆU ---")
display(df.head())  # Trong Jupyter Notebook dùng display() để hiện bảng cực đẹp!
# 2. Kiểm tra kiểu dữ liệu và số lượng ô không bị rỗng (Non-Null)
print("\n--- THÔNG TIN KIỂU DỮ LIỆU (INFO) ---")
df.info()
# 3. Đếm số lượng giá trị bị khuyết thiếu (NULL / NaN) ở từng cột
print("\n--- SỐ LƯỢNG Ô BỊ RỖNG Ở MỖI CỘT ---")
print(df.isna().sum())

print("\n--- THỐNG KÊ MÔ TẢ (DESCRIBE) ---")
# 4. Thống kê mô tả (describe) cho các cột số
print(df.describe())
```

```python
# 1. Xem các loại giao dịch và số lần xuất hiện
print("--- CỘT TRANSACTION_TYPE (ĐANG BỊ LỖN XỘN HOA/THƯỜNG) ---")
print(df["transaction_type"].value_counts())

# 2. Xem các trạng thái giao dịch
print("\n--- CỘT STATUS ---")
print(df["status"].value_counts())

# 3. Đếm xem có bao nhiêu khách hàng khác nhau trong bảng
print("\nSố lượng khách hàng duy nhất:", df["customer_id"].nunique())
```

```python
# 1. iloc: Lấy theo vị trí số (hàng từ 0 đến 3, cột từ 0 đến 4)
print("--- TRUY XUẤT BẰNG ILOC (THEO VỊ TRÍ SỐ NGUYÊN) ---")
display(df.iloc[0:4, 0:5])

# 2. loc: Lấy theo tên nhãn cột ('customer_name', 'amount', 'status')
print("\n--- TRUY XUẤT BẰNG LOC (THEO TÊN CỘT) ---")
display(df[["customer_name", "amount", "status"]].iloc[1:4])
```

```python
# 1. Lọc các giao dịch có trạng thái là 'FAILED'
failed_txns = df[df["status"] == "FAILED"]
print(f"Số lượng giao dịch thất bại (FAILED): {len(failed_txns)}")
display(failed_txns[["transaction_id", "customer_name", "amount", "branch_code"]])

# 2. Lọc theo danh sách chi nhánh dùng .isin() (Tương đương IN ('HN_01', 'HCM_01') trong SQL)
hn_hcm_txns = df[df["branch_code"].isin(["HN_01", "HCM_01"])]
print(f"\nSố lượng giao dịch thuộc chi nhánh HN_01 hoặc HCM_01: {len(hn_hcm_txns)}")
```

```python
# --- BƯỚC 1: XÓA DÒNG TRÙNG LẶP (DUPLICATES) ---
so_dong_trung = df.duplicated(subset=["transaction_id"]).sum()
print(f"Số dòng bị trùng transaction_id: {so_dong_trung}")

# Xóa bỏ dòng trùng, giữ lại dòng xuất hiện đầu tiên (keep='first')
df = df.drop_duplicates(subset=["transaction_id"], keep="first")
print(f"Kích thước bảng sau khi xóa trùng: {df.shape}")


# --- BƯỚC 2: CHUẨN HÓA CỘT TEXT (XÓA KHOẢNG TRẮNG & CHỮ HOA/THƯỜNG) ---
# Chuẩn hóa tên khách hàng: xóa khoảng trắng 2 đầu (.strip) và viết hoa chữ cái đầu (.title)
df["customer_name"] = df["customer_name"].str.strip().str.title()

# Chuẩn hóa loại giao dịch: viết hoa toàn bộ (.upper) và xóa khoảng trắng
df["transaction_type"] = df["transaction_type"].str.strip().str.upper()

# Chuẩn hóa mã chi nhánh: viết hoa (đổi hcm_02 thành HCM_02)
df["branch_code"] = df["branch_code"].str.strip().str.upper()

# Chuẩn hóa trạng thái: đưa 'Completed' và 'success' về chung 1 chữ 'SUCCESS'
df["status"] = df["status"].str.strip().str.upper()
df["status"] = df["status"].replace({"COMPLETED": "SUCCESS"})

print("\n--- KẾT QUẢ CỘT STATUS SAU KHI CHUẨN HÓA ---")
print(df["status"].value_counts())
```

```python
# --- BƯỚC 1: LÀM SẠCH CỘT AMOUNT ---
# 1. Bỏ ký tự '$' và dấu phẩy ',' bằng .str.replace
df["amount"] = df["amount"].astype(str).str.replace("$", "", regex=False).str.replace(",", "", regex=False).str.strip()

# 2. Ép sang số thực bằng pd.to_numeric với errors='coerce'
# MẸO CỦA DE: errors='coerce' sẽ tự động biến các chữ rác như 'N/A' thành NaN mà không bị văng lỗi crash!
df["amount"] = pd.to_numeric(df["amount"], errors="coerce")


# --- BƯỚC 2: ÉP KIỂU CỘT ACCOUNT_BALANCE ---
# Bỏ dấu phẩy (nếu có) và ép sang số thực
df["account_balance"] = df["account_balance"].astype(str).str.replace(",", "", regex=False).str.strip()
df["account_balance"] = pd.to_numeric(df["account_balance"], errors="coerce")

print("--- KIỂM TRA KIỂU DỮ LIỆU SAU KHI ÉP SỐ ---")
print(df[["amount", "account_balance"]].dtypes)
print("\n5 dòng đầu của cột số tiền:")
display(df[["amount", "account_balance"]].head())
```

```python
# 1. Dùng pd.to_datetime với tham số format='mixed'
# format='mixed' (tính năng từ Pandas 2.0+) tự động nhận diện thông minh nhiều định dạng ngày khác nhau!
df["transaction_date"] = pd.to_datetime(df["transaction_date"], format="mixed", errors="coerce")

# 2. Trích xuất thêm các cột thời gian phục vụ phân tích (dùng thuộc tính .dt)
df["tx_year"] = df["transaction_date"].dt.year            # Lấy Năm
df["tx_month"] = df["transaction_date"].dt.month          # Lấy Tháng
df["tx_day_name"] = df["transaction_date"].dt.day_name()  # Lấy Thứ trong tuần (Monday, Tuesday...)

print("--- KIỂM TRA CỘT NGÀY THÁNG ĐÃ ĐƯỢC CHUẨN HÓA ---")
display(df[["transaction_date", "tx_year", "tx_month", "tx_day_name"]].head())
```

```python
print("--- SỐ LƯỢNG Ô BỊ NULL HIỆN TẠI ---")
print(df.isna().sum())

# 1. Điền giá trị mặc định cho cột branch_code nếu bị rỗng hoặc là 'UNKNOWN'
df["branch_code"] = df["branch_code"].fillna("UNKNOWN")
df["branch_code"] = df["branch_code"].replace({"": "UNKNOWN"})

# 2. Điền số tiền amount bị thiếu bằng giá trị trung vị (Median) của cột
median_amount = df["amount"].median()
df["amount"] = df["amount"].fillna(median_amount)

# 3. Điền số dư account_balance bị thiếu bằng 0
df["account_balance"] = df["account_balance"].fillna(0.0)

print("\n--- SỐ LƯỢNG NULL SAU KHI XỬ LÝ (CHỈ CÒN THIẾU NGÀY GIAO DỊCH) ---")
print(df.isna().sum())
```

```python
# 1. Dùng np.where để phân loại giao dịch (tương tự hàm IF trong Excel hoặc CASE WHEN trong SQL)
# Nếu amount >= 5000$ thì đánh dấu là 'VIP_HIGH_VALUE', ngược lại là 'STANDARD'
df["transaction_tier"] = np.where(df["amount"] >= 5000, "VIP_HIGH_VALUE", "STANDARD")

# 2. Đánh dấu các giao dịch có số dư tài khoản bị âm (Dấu hiệu nợ xấu hoặc lỗi hệ thống)
df["is_negative_balance"] = df["account_balance"] < 0

# 3. Sắp xếp toàn bộ bảng giảm dần theo số tiền giao dịch (amount)
df_sorted = df.sort_values(by="amount", ascending=False)

print("--- TOP 5 GIAO DỊCH CÓ SỐ TIỀN LỚN NHẤT ---")
display(df_sorted[["transaction_id", "customer_name", "amount", "transaction_tier", "status"]].head(5))
```

```python
# 1. Báo cáo theo Chi nhánh (branch_code):
# Tính: Tổng số tiền (sum), Số lượng giao dịch (count), Số tiền trung bình (mean)
branch_report = df.groupby("branch_code")["amount"].agg(
    tong_tien="sum",
    so_giao_dich="count",
    trung_binh="mean"
).reset_index()

print("--- BÁO CÁO DOANH SỐ THEO CHI NHÁNH ---")
display(branch_report.sort_values(by="tong_tien", ascending=False))

# 2. Gom nhóm theo 2 cột: Loại giao dịch & Trạng thái
type_status_report = df.groupby(["transaction_type", "status"])["amount"].sum().unstack().fillna(0)
print("\n--- BẢNG TỔNG HỢP THEO LOẠI GIAO DỊCH & TRẠNG THÁI ---")
display(type_status_report)
```

```python
# 1. Giả lập một bảng thông tin chi nhánh (Dimension Table)
branch_info = pd.DataFrame({
    "branch_code": ["HN_01", "HN_02", "HCM_01", "HCM_02", "HCM_03", "DN_01", "DN_02"],
    "city": ["Hà Nội", "Hà Nội", "TP.HCM", "TP.HCM", "TP.HCM", "Đà Nẵng", "Đà Nẵng"],
    "branch_manager": ["Mr. Hoàng", "Ms. Lan", "Mr. Tuấn", "Ms. Mai", "Mr. Hùng", "Ms. Trang", "Mr. Long"]
})

# 2. Ghép nối bảng (LEFT JOIN theo cột 'branch_code')
df_final = pd.merge(df, branch_info, on="branch_code", how="left")

# Điền 'Chưa phân bổ' cho những giao dịch có chi nhánh UNKNOWN
df_final["city"] = df_final["city"].fillna("Chưa phân bổ")
df_final["branch_manager"] = df_final["branch_manager"].fillna("Chưa chỉ định")

print("--- 5 DÒNG DỮ LIỆU HOÀN THIỆN SAU KHI JOIN ---")
display(df_final[["transaction_id", "customer_name", "amount", "city", "branch_manager"]].head(5))


# 3. XUẤT KẾT QUẢ SẠCH RA FILE DÙNG CHO PIPELINE (EXPORT I/O):
# A. Xuất ra file CSV sạch (nhớ index=False)
df_final.to_csv("clean_bank_transactions.csv", index=False, encoding="utf-8")

# B. Xuất ra file Parquet chuẩn Data Engineer (Nén snappy, nhẹ và đọc siêu tốc)
df_final.to_parquet("clean_bank_transactions.parquet", compression="snappy", index=False)

print("\n-> CHÚC MỪNG BẠN! Đã làm sạch toàn bộ dữ liệu và lưu thành công 2 file:")
print("   1. clean_bank_transactions.csv")
print("   2. clean_bank_transactions.parquet")
```
